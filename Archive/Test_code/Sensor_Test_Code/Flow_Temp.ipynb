```python
from labjack import ljm
import time

# Open first found T7
handle = ljm.openS("T7", "ANY", "ANY")

info = ljm.getHandleInfo(handle)
print(f"Connected to T7")
print(f"IP Address: {ljm.numberToIP(info[3])}")
print()

# Configure inputs
ljm.eWriteName(handle, "AIN2_RANGE", 10.0)
ljm.eWriteName(handle, "AIN3_RANGE", 10.0)

try:
    while True:
        v2 = ljm.eReadName(handle, "AIN2")
        v3 = ljm.eReadName(handle, "AIN3")

        # LJTCS conversion
        i2_mA = v2 * 8.475
        i3_mA = v3 * 8.475
        '''
        print(
            f"AIN2: {v2:6.4f} V ({i2_mA:5.2f} mA)   "
            f"AIN3: {v3:6.4f} V ({i3_mA:5.2f} mA)"
        )
        ''' 
        temp_mA = v2 * 8.475

        flow_mA = v3 * 8.475

        # Temperature
        temp_f = 4 + ((temp_mA - 4) / 16.0) * (176 - 4)
        temp_c = (temp_f - 32) * 5 / 9

        # Flow
        flow_gpm = ((flow_mA - 4) / 16.0) * 6.6
        flow_gpm = max(0, flow_gpm)

        print(
            f"Flow = {flow_gpm:.2f} GPM | "
            f"Temp = {temp_c:.1f} °C ({temp_f:.1f} °F)"
        )
        time.sleep(0.5)

except KeyboardInterrupt:
    print("\nStopping...")

finally:
    ljm.close(handle)
```
