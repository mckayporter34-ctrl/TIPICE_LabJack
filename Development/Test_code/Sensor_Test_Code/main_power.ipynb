```python
from labjack import ljm
import time

handle = ljm.openS("T7", "ANY", "ANY")

try:
    time.sleep(5)
    print("Setting FIO0 HIGH")
    ljm.eWriteName(handle, "FIO0", 1)

    time.sleep(5)

    print("Setting FIO0 LOW")
    ljm.eWriteName(handle, "FIO0", 0)

finally:
    ljm.close(handle)
```
